```python
import pandas as pd
import numpy as np

metrics = ['GED (Fidelity)', 'Latent Euclidean Distance (Fidelity)', 'Latent Cossimilary (Fidelity)', 'Absolute Difference (Validity)', 'flip-ratio (Validity)']
methods = ['random_sampled_CFs_list (mean)', 'closest_graph_train (mean)', 'closest_latent_train (mean)', 'main_method (mean)']

methods_name_map = {
    'random_sampled_CFs_list (mean)': 'Random',
    'closest_latent_train (mean)': 'Decoded Mean of k-NN',
    'closest_graph_train (mean)': 'Graph of NN from Training',
    'main_method (mean)': 'Classifier Guided CF',
}

metric_name_map = {
    'GED (Fidelity)': 'GED (Fidelity) \\downarrow',
    'Latent Euclidean Distance (Fidelity)': 'Latent Euclidean Distance (Fidelity) \\downarrow',
    'Latent Cossimilary (Fidelity)': 'Latent Cossimilary (Fidelity) \\uparrow',
    'Absolute Difference (Validity)': 'Absolute Difference (Validity) \\uparrow',
    'flip-ratio (Validity)': 'flip-ratio (Validity) \\uparrow'
}

latex_config = {
    'column_format': 'c|ccc||cc',
    'float_format': "%.2f",
}

def configure_dataframes(path,):
    df = pd.read_csv(path, index_col=0)
    df_t = df[methods].rename(columns=methods_name_map).transpose()
    for metric in metrics:
        if not metric in df_t.keys():
            df_t[metric] = np.nan
    #filtered_metrics = filter(lambda x: x in df.transpose().keys(), metrics)
    return df_t[metrics].round(decimals=2).rename(columns=metric_name_map)

def format_mean_std(mean, std):
    if pd.isna(std):  # Check if std is NaN
        return f"{mean}"  # Only return the mean without ±
    else:
        return f"{mean} $\\pm$ {{\\small {std}}}"  # Return mean ± std

def get_result_dataframe(df_mean, df_std):
    results = pd.DataFrame(
        {col: [format_mean_std(m, s) for m, s in zip(df_mean[col], df_std[col])] for col in df_mean.columns},
        index=df_mean.index
    )
    latex_results = results.to_latex(**latex_config)
    return results, latex_results
```

## Aids

```python
# Paths
method = 'test_aids'
path_mean = f'./data/predictions/{method}_method_statistics/aggregated_mean_counterfactual_statistics.csv'
path_std = f'./data/predictions/{method}_method_statistics/aggregated_std_counterfactual_statistics.csv'

results_aids, latex_results = get_result_dataframe(configure_dataframes(path_mean), configure_dataframes(path_std))
results_aids
```

```python
print(latex_results)
```
